````python
from rdflib import Graph, URIRef, Literal, RDF
from rdflib.namespace import FOAF, XSD

g = Graph()

print(g)
print("Number of triples:", len(g))
````

````python
donna = URIRef("http://example.org/donna")

g.add((donna, RDF.type, FOAF.Person))
g.add((donna, FOAF.name, Literal("Donna Fales")))
g.add((donna, FOAF.nick, Literal("donna", lang="en")))
g.add((donna, FOAF.mbox, URIRef("mailto:donna@example.org")))

print("Number of triples:", len(g))
````

````python
for s, p, o in g:
    print("SUBJECT:  ", s)
    print("PREDICATE:", p)
    print("OBJECT:   ", o)
    print()
````

````python
ed = URIRef("http://example.org/edward")

g.add((ed, RDF.type, FOAF.Person))
g.add((ed, FOAF.name, Literal("Edward Scissorhands")))
g.add((ed, FOAF.nick, Literal("ed", datatype=XSD.string)))
g.add((ed, FOAF.mbox, Literal("e.scissorhands@example.org", datatype=XSD.anyURI)))

print("Number of triples:", len(g))
````

````python
for person in g.subjects(RDF.type, FOAF.Person):
    print(person)
````

````python
for person in g.subjects(RDF.type, FOAF.Person):
    print("Person:", person)
    
    for mailbox in g.objects(person, FOAF.mbox):
        print("Mailbox:", mailbox)
    
    print()
````

````python
g.bind("foaf", FOAF)

print(g.serialize(format="turtle"))
````

````python
q = """
PREFIX foaf: <http://xmlns.com/foaf/0.1/>

SELECT ?person ?name
WHERE {
    ?person a foaf:Person .
    ?person foaf:name ?name .
}
"""

for row in g.query(q): 
    print(row.person, "=>", row.name)
````

## Challenge 1: Add yourself and print only nicknames

Create three people: you, Donna, and Edward. Give each one a `foaf:name` and a `foaf:nick`.

Then print only the nicknames.

Starter:

```python
from rdflib import Graph, URIRef, Literal, RDF
from rdflib.namespace import FOAF

g = Graph()
g.bind("foaf", FOAF)

me = URIRef("http://example.org/me")
donna = URIRef("http://example.org/donna")
edward = URIRef("http://example.org/edward")

g.add((me, RDF.type, FOAF.Person))
g.add((me, FOAF.name, Literal("Your Name")))
g.add((me, FOAF.nick, Literal("your_nick")))

g.add((donna, RDF.type, FOAF.Person))
g.add((donna, FOAF.name, Literal("Donna Fales")))
g.add((donna, FOAF.nick, Literal("donna")))

g.add((edward, RDF.type, FOAF.Person))
g.add((edward, FOAF.name, Literal("Edward Scissorhands")))
g.add((edward, FOAF.nick, Literal("ed")))
```

Your task:

```python
for person in g.subjects(RDF.type, FOAF.Person):
    # print the person's nickname
    ...
```

Expected idea:

```text
your_nick
donna
ed
```

Hint:

```python
g.objects(person, FOAF.nick)
```

## Challenge 2: Add friendship links

Use `FOAF.knows` to say:

```text
You know Donna.
Donna knows Edward.
Edward knows you.
```

Add these triples:

```python
g.add((me, FOAF.knows, donna))
g.add((donna, FOAF.knows, edward))
g.add((edward, FOAF.knows, me))
```

Now print sentences like this:

```text
http://example.org/me knows http://example.org/donna
http://example.org/donna knows http://example.org/edward
http://example.org/edward knows http://example.org/me
```

Your task:

```python
for person, friend in g.subject_objects(FOAF.knows):
    print(person, "knows", friend)
```

Bonus: print names instead of ugly URLs.

Hint function:

```python
def get_name(graph, person):
    for name in graph.objects(person, FOAF.name):
        return name
    return person
```

Then:

```python
for person, friend in g.subject_objects(FOAF.knows):
    print(get_name(g, person), "knows", get_name(g, friend))
```

Expected output:

```text
Your Name knows Donna Fales
Donna Fales knows Edward Scissorhands
Edward Scissorhands knows Your Name
```

## Challenge 3: Query with SPARQL

Now use SPARQL to ask:

> Give me every person and their nickname.

```python
q = """
PREFIX foaf: <http://xmlns.com/foaf/0.1/>

SELECT ?person ?nick
WHERE {
    ?person a foaf:Person .
    ?person foaf:nick ?nick .
}
"""

for row in g.query(q):
    print(row.person, "=>", row.nick)
```

Now modify it to get names too:

```sparql
SELECT ?person ?name ?nick
WHERE {
    ?person a foaf:Person .
    ?person foaf:name ?name .
    ?person foaf:nick ?nick .
}
```

Expected output shape:

```text
http://example.org/me => Your Name => your_nick
http://example.org/donna => Donna Fales => donna
http://example.org/edward => Edward Scissorhands => ed
```

---

The secret muscle you are training is this:

```text
subject      predicate       object
me           foaf:name       "Your Name"
me           foaf:nick       "your_nick"
me           foaf:knows      donna
donna        foaf:knows      edward
```

Once that pattern clicks, RDF becomes much less “what is this alien spaghetti” and more “tiny facts connected by labeled arrows.”
